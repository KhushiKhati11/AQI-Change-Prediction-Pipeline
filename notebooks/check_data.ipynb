```python
import pandas as pd
import matplotlib.pyplot as plt

teku = pd.read_csv("../data/raw/teku.csv")
mid_b = pd.read_csv("../data/raw/mid_baneshwor.csv")

#The start and end date of the data
print("Teku:", teku.shape, teku["datetime_utc"].min(), teku["datetime_utc"].max())
print("Mid Baneshwor:", mid_b.shape, mid_b["datetime_utc"].min(), mid_b["datetime_utc"].max())
```

```python
#Check for number of missing values
print(teku.isna().sum())
print(mid_b.isna().sum())
```

```python
# Calculate expected hours: Max datetime - Min datetime. Since the data is hourly, we can calculate the expected number of hours by taking the difference in days and multiplying by 24

expected_hours = 24 * (pd.to_datetime("2026-09-04") - pd.to_datetime("2025-11-25")).days
print(f"Teku completeness: {len(teku)/expected_hours:.1%}")
print(f"Mid Baneshwor completeness: {len(mid_b)/expected_hours:.1%}")
```

```python
# Plotting the raw data coverage for Teku PM2.5 (Line graph)

teku["datetime_utc"] = pd.to_datetime(teku["datetime_utc"])        # Convert datetime_utc to datetime object
teku.plot(x="datetime_utc", y="pm25", figsize=(15, 4), title="Teku PM2.5 - Raw Coverage")
plt.show()
```

```python
""" Calculate the full range of datetime values in the Teku dataset, with an hourly frequency. Then, reindex the Teku DataFrame to this full range,
which will introduce NaN values for any missing hours. Finally, plot the PM2.5 values to visualize the gaps in the data.
"""
full_range = pd.date_range(start=teku["datetime_utc"].min(), end=teku["datetime_utc"].max(), freq="h")
teku_reindexed = teku.set_index("datetime_utc").reindex(full_range)
teku_reindexed["pm25"].plot(figsize=(15,4), title="Teku PM2.5 - true gaps visible")
```

```python
# Analyze the gaps in the reindexed Teku PM2.5 data

is_missing = teku_reindexed["pm25"].isna()                              # Identify missing values in the PM2.5 column
gap_groups = (is_missing != is_missing.shift()).cumsum()                #Create groups for consecutive missing values
gap_lengths = teku_reindexed[is_missing].groupby(gap_groups).size()     #Calculate the length of each gap
print(gap_lengths.describe())
gap_lengths.sort_values(ascending=False).head(10)                       #Print the lengths of the 10 longest gaps in the Teku PM2.5 data
```

```python
gap_start_times = teku_reindexed[is_missing].groupby(gap_groups).apply(lambda x: x.index[0])   # Get the start time of each gap
gaps_per_month = gap_start_times.dt.to_period("M").value_counts().sort_index()                 # Count the number of gaps per month and sort by month
print(gaps_per_month)
```

Now the same steps are repeated for Mid Baneshwor dataset.

```python
# Plotting the raw data coverage for Mid Baneshwor PM2.5

mid_b["datetime_utc"] = pd.to_datetime(mid_b["datetime_utc"])
mid_b.plot(x="datetime_utc", y="pm25", figsize=(15, 4), title="Mid-Baneshwor PM2.5 - Raw Coverage")
plt.show()
```

```python
full_range = pd.date_range(start=mid_b["datetime_utc"].min(), end=mid_b["datetime_utc"].max(), freq="h")
mid_b_reindexed = mid_b.set_index("datetime_utc").reindex(full_range)
mid_b_reindexed["pm25"].plot(figsize=(15,4), title="Mid-B PM2.5 - true gaps visible")
```

```python
is_missing = mid_b_reindexed["pm25"].isna()
gap_groups = (is_missing != is_missing.shift()).cumsum()
gap_lengths = mid_b_reindexed[is_missing].groupby(gap_groups).size()
print(gap_lengths.describe())
print(gap_lengths.sort_values(ascending=False).head(10))
```

```python
gap_start_times = mid_b_reindexed[is_missing].groupby(gap_groups).apply(lambda x: x.index[0])
gaps_per_month = gap_start_times.dt.to_period("M").value_counts().sort_index()
print(gaps_per_month)
```
